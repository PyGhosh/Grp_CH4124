```python
test_set = ["CCO", "CC(=O)O", "ClCCBr", "c1ccncc1", "OS(=O)(=O)O"]
def heavy_atoms(smiles):
  counts = {}
  for s in smiles:
    heavy_a_count=0
    i=0
  while i<len(s):
    char=s[i]
    if char=='C' and (i+1)<len(s) and s[i+1]=='l':
      heavy_a_count+=1
      i+=2
    elif char=='B' and (i+1)<len(s) and s[i+1]=='r':
      heavy_a_count+=1
      i+=2
    elif char in ['C','c','N','O','S','F','P','I','B']:
      heavy_a_count+=1
      i+=1
    else:
      i+=1
    counts[s]=heavy_a_count
    pass
  return counts
for idx in test_set:
  print(idx, heavy_atoms(idx))
```

```python
eps=np.array([[1.0, 0.5],[0.2,1.0]])
A=np.array([1.2,0.8])

det_eps=np.linalg.det(eps)
print(det_eps)

conc=np.linalg.solve(eps,A)
print(conc)

prod=np.prod(conc)
print(prod)
```

```python
A6=np.zeros((6,6))
for i in range(0,6):
  if i+1<6:
    A6[i][i+1]=1
    A6[i+1][i]=1
eigenvalues=np.linalg.eigvalsh(A6)
eigenvalues=np.sort(eigenvalues)[::-1]
print(eigenvalues)

occupied_eigenvalues = eigenvalues[:3]
E_pi_coeff_beta = 2 * np.sum(occupied_eigenvalues)
E_localized_coeff_beta = 6.0

delocal_energy_coeff_beta = E_pi_coeff_beta - E_localized_coeff_beta

print(f"E_pi = {E_pi_coeff_beta:.4f} beta")
print(f"Delocalization Energy = {delocal_energy_coeff_beta:.4f} beta")
```

```python
T=np.array([300.,320.,340.,360.,380.])
k=np.array([1.71e-3, 6.03e-3, 1.83e-2, 4.92e-2, 1.19e-1])
R=8.314
ln_k = np.log(k)
inverse_T = 1/T
slope, intercept = np.polyfit(inverse_T, ln_k, 1)
Ea = -slope * R
A = np.exp(intercept)
Ea_kJ = Ea / 1000
ln_k_pred = slope * inverse_T + intercept
resi = ln_k - ln_k_pred
largest_abs_resi = np.max(np.abs(resi))
print(f"Slope: {slope:.2f}")
print(f"Intercept: {intercept:.2f}")
print(f"Activation Energy: {Ea_kJ:.2f} kJ/mol")
print(f"Pre-exponential Factor: {A:.2e}")
print(f"Largest Absolute Residual in ln k: {largest_abs_resi:.2e}")
```
