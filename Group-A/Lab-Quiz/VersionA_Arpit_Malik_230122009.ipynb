```python
import numpy as np
```

```python
test_set = ["CCO" , "CC(=O)O" , "ClCCBr" , "c1ccncc1" , "OS(=O)(=O)O"]
```

```python
def heavy_atoms(smiles_list):
  counts = {}
  for s in smiles_list:
    heavy_atom_count = 0
    idx = 0
    while idx < len(s):
      char = s[idx]
      if char == 'C' and (idx + 1) < len(s) and s[idx+1] == 'l':
        heavy_atom_count += 1
        idx += 2
      elif char == 'B' and idx + 1 < len(s) and s[idx+1] =='r':
        heavy_atom_count += 1
        idx += 2
      elif char in ['C','c','O','N', 'S','F','P','I', 'B']:
        heavy_atom_count += 1
        idx += 1
      else:
        idx += 1
    counts[s] = heavy_atom_count
  return counts
```

```python
for i in test_set:
  print(i ,"Toatal Heavy atoms = ", heavy_atoms([i]))
```

```python
A6 = np.zeros((6,6))
for i in range(0,6):
  if i+1 < 6 :
    A6[i][i+1] = 1
    A6[i+1][i] = 1
print(A6)


eigenvalues  = np.linalg.eigvalsh(A6)
print("Eigenvalues :", eigenvalues)


eigenvalues_descending = np.sort(eigenvalues)[::-1]
print("Eigenvalues descending:", eigenvalues_descending)

delocalisation_energy = eigenvalues[0] - eigenvalues[1]
print("delocalisation_energy:", delocalisation_energy)
```
