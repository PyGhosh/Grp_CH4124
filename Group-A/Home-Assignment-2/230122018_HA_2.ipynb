## Task 2. Colour refinement  &nbsp;&nbsp;

**Do this by hand first, on paper.** The code is to check your hand work, not to replace it.

The 1-WL refinement rule is

$$c^{(k+1)}_i \;=\; \mathrm{HASH}\Big(c^{(k)}_i,\ \{\!\{\,c^{(k)}_j : j \in \mathcal{N}(i)\,\}\!\}\Big)$$

where $\{\!\{\cdot\}\!\}$ is a **multiset** (repetition matters, order does not) and
$\mathrm{HASH}$ assigns a fresh integer to each distinct signature it has seen.

Two graphs are declared **distinguishable** if at some round their *multisets of colours*
differ.

Complete the function below.


````python
def wl_round(A, colours):
    """One round of 1-WL colour refinement.

    A       : (n, n) adjacency matrix
    colours : list of n integers, the current colours
    returns : list of n integers, the refined colours
    """
    n = len(colours)
    signatures = []
    for i in range(n):
        # TODO: build the multiset of neighbour colours for atom i.
        #       Use sorted(...) to make the multiset order-independent,
        #       and tuple(...) so it can be used as a dictionary key.
        neighbour_colours = tuple(sorted(colours[j] for j in range(n) if A[i, j] == 1))

        # TODO: the signature is the pair (own colour, neighbour multiset)
        signatures.append((colours[i], neighbour_colours))

    # relabel each distinct signature with a fresh integer (this part is done for you)
    table = {s: k for k, s in enumerate(sorted(set(signatures), key=str))}
    return [table[s] for s in signatures]
````

## Task 4. Find a discriminating invariant  &nbsp;&nbsp;

Refinement failed. Something else must succeed, because the two graphs genuinely differ.

Recall the walk-counting theorem: $(A^k)_{ij}$ is the number of walks of length exactly
$k$ from atom $i$ to atom $j$. Therefore $\mathrm{tr}(A^k) = \sum_i (A^k)_{ii}$ counts
**closed** walks of length $k$, that is walks that return to where they started.

### YOUR ANSWER (Task 4)

(a) Which value of $k$ separates A from B?

(b) Explain **in terms of walks** why that particular $k$ works. What closed walk exists
in one graph and not the other?

(c) Explain why the other two values of $k$ fail. Be careful with $k=6$: the result may
not be what you expected, and the explanation is the point of this part.

(d) One sentence: what does $\mathrm{tr}(A^2)$ count for *any* graph, and why could it

*Write here:*

(a) **$k = 3$** separates A from B: $\mathrm{tr}(A_A^3) = 0$ while $\mathrm{tr}(A_B^3) = 12$.

(b) A closed walk of length 3 that returns to its start is, for a simple graph, exactly a
walk around a triangle (traversed in either direction, starting from any of its three
atoms). B contains two genuine triangles (the two cyclopropane rings), and each triangle
contributes $3 \times 2 = 6$ such closed 3-walks (3 choices of starting atom, 2 directions
around the ring), giving $\mathrm{tr}(A_B^3) = 12$. A (the six-membered ring) contains no
triangles at all — you cannot leave an atom, take three steps, and return to it around a
6-cycle — so $\mathrm{tr}(A_A^3) = 0$. The trace directly counts triangles, and only B has any.

(c) $k=2$ fails because $\mathrm{tr}(A^2)$ counts closed walks of length 2, which for a
simple undirected graph are just "walk along an edge and immediately walk back" — there
are exactly two such walks per edge (one in each direction), so $\mathrm{tr}(A^2) = 2|E|$
for *any* graph. Since A and B both have exactly 6 bonds, they necessarily give the same
value (12), no matter how those bonds are arranged into rings. $k=6$ fails for a much
more surprising reason: even though A is a single 6-ring (so it has closed 6-walks that go
exactly once around the whole molecule) and B is two 3-rings (so its 6-walks come from
going around a triangle twice, in various orders), these two very different walk-counting
stories happen to add up to the *same total*, $\mathrm{tr}(A_A^6) = \mathrm{tr}(A_B^6) = 132$.
This is visible in the eigenvalues: A's spectrum is $\{2, 1, 1, -1, -1, -2\}$ and B's is
$\{2, 2, -1, -1, -1, -1\}$ — genuinely different multisets — but $\mathrm{tr}(A^6) = \sum_i
\lambda_i^6$, and raising to the 6th power erases the sign differences between $\pm 1$ and
$\pm 2$, so the two different spectra happen to sum to the same sixth-power total. This is
a good reminder that "a bigger $k$" does not monotonically give "more discriminating
power" — it can just as easily wash out a difference that a smaller $k$ revealed.

(d) $\mathrm{tr}(A^2)$ counts twice the number of edges in the graph ($2|E|$, from walking
out along an edge and immediately back), so it is a global edge-count statistic that
carries no information about how those edges are arranged into rings — it could never
separate A from B because both graphs have exactly the same number of bonds.

## Task 5. Propose a fix and defend it  &nbsp;&nbsp;

Computing $\mathrm{tr}(A^k)$ costs $O(n^3)$ and does not transfer cleanly between
molecules of different size. A cheaper repair is to give each atom an extra **input
feature** that already distinguishes the two cases, so that the model separates them at
layer zero without any change to the architecture.

Implement your chosen feature below.


````python
def extra_feature(A):
    """Return a length-n array: one extra scalar feature per atom.

    TODO: choose a feature that (i) differs between A and B,
          (ii) is computable in linear time by any cheminformatics toolkit,
          (iii) is chemically meaningful, not an arbitrary code.
    """
    n = A.shape[0]
    feature = np.zeros(n)

    # TODO: fill in
    # Feature: the size of the smallest ring that atom i belongs to
    # (0 if the atom is not part of any ring).
    #
    # This is exactly the "ring size" descriptor computed by every
    # cheminformatics toolkit's ring-perception step (e.g. RDKit's
    # GetRingInfo().AtomRingSizes(), built on an SSSR search).
    for i in range(n):
        neighbours = [j for j in range(n) if A[i, j] == 1]
        best = np.inf
        for start in neighbours:
            # Remove the edge (i, start) and look for the shortest path
            # back from `start` to `i`; that path length + 1 is the size
            # of the ring that closes through this particular neighbour.
            A_cut = A.copy()
            A_cut[i, start] = A_cut[start, i] = 0

            # simple BFS shortest-path length from `start` to `i`
            visited = {start}
            frontier = [start]
            dist = {start: 0}
            found = None
            while frontier:
                nxt = []
                for u in frontier:
                    for v in range(n):
                        if A_cut[u, v] == 1 and v not in visited:
                            visited.add(v)
                            dist[v] = dist[u] + 1
                            if v == i:
                                found = dist[v]
                            nxt.append(v)
                if found is not None:
                    break
                frontier = nxt

            if found is not None:
                best = min(best, found + 1)

        feature[i] = 0 if best == np.inf else best

    return feature

fA, fB = extra_feature(A_A), extra_feature(A_B)
print('feature on A:', fA)
print('feature on B:', fB)
````

*Write here:*

(a) The feature is **ring size** — the number of atoms in the smallest ring that contains
each atom. On A (cyclohexane), every atom sits in the single six-membered ring, so the
feature value is 6 for all six atoms. On B (two cyclopropanes), every atom sits in a
three-membered ring, so the feature value is 3 for all six atoms. Since $\{6,6,6,6,6,6\}
\neq \{3,3,3,3,3,3\}$ as multisets, this immediately separates A from B, before any
message passing happens.

(b) Ring size is computed by a **ring-perception algorithm** (the smallest set of smallest
rings, SSSR), the same routine that underlies `RDKit`'s `GetRingInfo()` or `OpenBabel`'s
ring perception. For real molecules — where every atom has a small, bounded valence (at
most 4 for carbon) and rings are typically small — this is effectively **linear in the
number of atoms**, because the shortest-cycle search from each atom only has to explore a
small local neighbourhood before it closes back on itself. It is far cheaper than the
$O(n^3)$ cost of forming $A^k$ and taking its trace, and unlike the trace it is defined
*per atom*, so it plugs straight into the existing node-feature vector.

(c) Ring size is one of the strongest predictors of **ring strain energy**: three- and
four-membered rings (like B's cyclopropanes) carry tens of kJ/mol of angle strain because
their internal angles are forced far from the ideal ~109.5° tetrahedral angle, whereas
six-membered rings (like A) can adopt a strain-free chair conformation. A model given
ring size as an input feature would therefore have direct access to a strong physical
signal for predicting strain energy, reactivity toward ring-opening, and even NMR
chemical shifts, all of which correlate strongly with ring size.

(d) When a model provably cannot see a distinction — because the *architecture itself*,
not the training procedure, is blind to it — the correct fix is a **better input feature**,
not a bigger network: no amount of extra width, depth, or data can create a signal that
message passing is mathematically incapable of propagating, but a single well-chosen
feature can restore it at negligible cost.
