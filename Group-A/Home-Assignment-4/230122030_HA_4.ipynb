## Exercise 1. Argon dimer in real units and the finite-difference step
Lennard-Jones argon: $\epsilon=0.0104$ eV, $\sigma=3.40$ Å,
$$U(r)=4\epsilon\big[(\sigma/r)^{12}-(\sigma/r)^6\big],\qquad F(r)=\frac{24\epsilon}{r}\big[2(\sigma/r)^{12}-(\sigma/r)^6\big].$$
The central difference has a **truncation error** $\propto h^2$ (too large $h$) and a **round-off error** $\propto\varepsilon_{\rm mach}/h$ (too small $h$). We scan $h$ to find the sweet spot.

```python
import numpy as np, matplotlib.pyplot as plt

eps, sig = 0.0104, 3.40                                   # eV, Angstrom
U = lambda r: 4 * eps * ((sig / r)**12 - (sig / r)**6)
F = lambda r: 24 * eps / r * (2 * (sig / r)**12 - (sig / r)**6)

r = 3.6                                                   # Angstrom, on the repulsive side of the minimum
print(f"U({r}) = {1000*U(r):.3f} meV,  F({r}) = {F(r):+.5f} eV/A")

hs = np.logspace(-12, -1, 45)
err = [abs(-(U(r + h) - U(r - h)) / (2 * h) - F(r)) for h in hs]
best = hs[np.argmin(err)]
print(f"best step h = {best:.1e} A, error = {min(err):.1e} eV/A")

plt.loglog(hs, err, "o-", ms=3)
plt.xlabel("finite-difference step h (A)"); plt.ylabel("|F_numeric - F_analytic| (eV/A)")
plt.title("round-off (left) versus truncation (right)"); plt.show()
```

**Your turn.** Theory predicts the best step near $h\sim\varepsilon_{\rm mach}^{1/3}\times(\text{length scale})$ with $\varepsilon_{\rm mach}\approx2\times10^{-16}$. Does your optimum agree? Repeat with the forward difference $-[U(r+h)-U(r)]/h$. Where is its optimum, and why is it worse?

```python
errf = [abs(-(U(r + h) - U(r)) / h - F(r)) for h in hs]
bestf = hs[np.argmin(errf)]
print(f"forward: best step h = {bestf:.1e} A, error = {min(errf):.1e} eV/A")
print(f"rule of thumb eps_mach^(1/3) * r = {np.finfo(float).eps**(1/3) * r:.1e} A")

plt.loglog(hs, err, "o-", ms=3, label="central"); plt.loglog(hs, errf, "o-", ms=3, label="forward")
plt.xlabel("finite-difference step h (A)"); plt.ylabel("|F_numeric - F_analytic| (eV/A)")
plt.title("round-off (left) versus truncation (right)"); plt.legend(); plt.show()
```

Answer. For the forward difference, the truncation error is first order, about h times U’’ divided by 2. The optimum moves to h of roughly the square root of eps_mach, about 1e-8 times the length scale, and the best achievable error is about the square root of eps_mach. The scan gave h = 1e-8 A with error 5.5e-10 eV/A, which matches this. That is roughly four orders of magnitude worse than the central difference. The central difference cancels the first-order error term by symmetry, which is why it does better.

## Exercise 2. Curvature of three cutoff functions
For energy conservation, the force (first derivative) must be continuous. For vibrational properties and stable integration, the **second** derivative should also be smooth. Compare at $r_c=5$ Å
- cosine: $\tfrac12[\cos(\pi r/r_c)+1]$,
- MTP-style: $(1-r/r_c)^2$,
- polynomial envelope with $p=6$.

```python
import numpy as np, matplotlib.pyplot as plt

rc = 5.0
r = np.linspace(0, 5.5, 5501); x = np.minimum(r / rc, 1)
cuts = {
    "cosine":       np.where(r < rc, 0.5 * (np.cos(np.pi * x) + 1), 0),
    "(1 - r/rc)^2": np.where(r < rc, (1 - x)**2, 0),
    "envelope p=6": np.where(r < rc, 1 - 28 * x**6 + 48 * x**7 - 21 * x**8, 0),
}
dr = r[1] - r[0]
fig, ax = plt.subplots(1, 3, figsize=(12, 3))
for name, f in cuts.items():
    d1 = np.gradient(f, dr); d2 = np.gradient(d1, dr)
    ax[0].plot(r, f, label=name); ax[1].plot(r, d1, label=name); ax[2].plot(r, d2, label=name)
    i = np.searchsorted(r, rc) - 5
    print(f"{name:14s}: just inside rc  f = {f[i]:.1e},  f' = {d1[i]:+.1e},  f'' = {d2[i]:+.3f}")
for a_, t in zip(ax, ("f", "f'", "f''")): a_.set_title(t); a_.axvline(rc, color="gray", ls=":"); a_.set_xlabel("r (A)")
ax[2].set_ylim(-0.3, 0.3); ax[0].legend(); plt.tight_layout(); plt.show()
```

**Your turn.** Rank the three functions by smoothness at $r_c$ (value, slope, curvature). The cosine and $(1-r/r_c)^2$ both have zero slope but nonzero curvature at $r_c$, so the forces are continuous while their derivatives jump. Which property (energy conservation in MD, or phonon frequencies from the Hessian) is affected by a curvature jump, and why?

Answer. From smoothest to roughest I get the envelope, then the MTP-style function, then the cosine. All three reach zero value and zero slope at rc, so the ranking comes down to curvature. The numbers printed just inside rc are about 0.20 for the cosine, 0.08 for the MTP-style function and 0.01 for the envelope, which is already heading to zero. So the cosine has the biggest jump.

Phonon frequencies from the Hessian are the property that suffers from a curvature jump. The Hessian is the matrix of second derivatives of the energy, and with the cutoff value and slope both zero at rc, a pair that crosses rc changes its force constant by the pair energy at rc times the curvature jump. Finite-displacement phonon runs move atoms by around 0.01 A, enough to push a neighbour shell sitting near rc in or out of the sum, so the frequencies can jump with tiny changes in strain. Energy conservation gets off lightly. Velocity Verlet only needs a continuous force to avoid a sudden kick, and all three cutoffs have zero slope at rc, so the force is continuous. A neighbour crossing the cutoff still makes the integration error a little larger on that one step, which stays small at normal time steps. For anything that touches the Hessian I would pick the envelope.

## Exercise 3. Radial fingerprint of a carbon atom in graphene
$$G^2_i(R_s)=\sum_{j\ne i}e^{-\eta(r_{ij}-R_s)^2}f_c(r_{ij}).$$
In graphene ($d_{\rm CC}=1.42$ Å) each carbon has 3 first neighbours at 1.42 Å, 6 second neighbours at 2.46 Å and 3 third neighbours at 2.84 Å. We build a patch of the honeycomb lattice and compute the fingerprint of one atom.

```python
import numpy as np, matplotlib.pyplot as plt

fc = lambda r, rc: np.where(r < rc, 0.5 * (np.cos(np.pi * r / rc) + 1), 0.0)

d = 1.42
a1, a2 = d * np.array([1.5, np.sqrt(3) / 2]), d * np.array([1.5, -np.sqrt(3) / 2])   # lattice vectors
basis = [np.zeros(2), np.array([d, 0.0])]                                           # two atoms per cell
atoms = np.array([i * a1 + j * a2 + b for i in range(-4, 5) for j in range(-4, 5) for b in basis])
centre = atoms[np.argmin(np.linalg.norm(atoms, axis=1))]
r0j = np.linalg.norm(atoms - centre, axis=1); r0j = r0j[r0j > 1e-6]
for s in np.unique(r0j[r0j < 3.0].round(3)):
    print(f"shell at {s:.3f} A: {np.sum(np.isclose(r0j, s, atol=1e-3))} atoms")

Rs = np.linspace(1.0, 4.0, 400)
for eta, rc in [(10.0, 4.0), (10.0, 6.0), (2.0, 6.0)]:
    G2 = [np.sum(np.exp(-eta * (r0j - s)**2) * fc(r0j, rc)) for s in Rs]
    plt.plot(Rs, G2, label=f"eta = {eta}, rc = {rc} A")
plt.xlabel("R_s (A)"); plt.ylabel("G2 of one carbon atom"); plt.legend(); plt.show()
```

**Your turn.** Stretch the lattice by 2 % (multiply `d` by 1.02) and plot the difference in $G^2$. Where does the fingerprint change most? This is how an MLIP \"sees\" strain.

```python
d2 = 1.02 * d                                                                        # stretch the lattice by 2 %
a1, a2 = d2 * np.array([1.5, np.sqrt(3) / 2]), d2 * np.array([1.5, -np.sqrt(3) / 2])
basis = [np.zeros(2), np.array([d2, 0.0])]
atoms = np.array([i * a1 + j * a2 + b for i in range(-4, 5) for j in range(-4, 5) for b in basis])
centre = atoms[np.argmin(np.linalg.norm(atoms, axis=1))]
r0j2 = np.linalg.norm(atoms - centre, axis=1); r0j2 = r0j2[r0j2 > 1e-6]

for eta, rc in [(10.0, 4.0), (10.0, 6.0), (2.0, 6.0)]:
    G2 = np.array([np.sum(np.exp(-eta * (r0j - s)**2) * fc(r0j, rc)) for s in Rs])
    G2s = np.array([np.sum(np.exp(-eta * (r0j2 - s)**2) * fc(r0j2, rc)) for s in Rs])
    print(f"eta = {eta}, rc = {rc} A: largest change {np.max(np.abs(G2s - G2)):.3f} at R_s = {Rs[np.argmax(np.abs(G2s - G2))]:.2f} A")
    plt.plot(Rs, G2s - G2, label=f"eta = {eta}, rc = {rc} A")
plt.xlabel("R_s (A)"); plt.ylabel("G2(stretched) - G2(original)"); plt.legend(); plt.show()
```

Answer. The fingerprint changes most around R_s of 2.3 A for eta of 10, and around 2.2 A for eta of 2. That is the inner flank of the second-shell peak at 2.46 A. The largest change printed is 0.34 for rc of 4 A and 0.59 for rc of 6 A, against peak heights of roughly 2 to 5 in the earlier plot, so a 2 percent stretch shows up as a change of 14 to 16 percent in the fingerprint. Part of it is the count. The second shell holds 6 atoms and each moves by 0.049 A, while the 3 first-shell atoms move only 0.028 A, so the first-shell lobes come out at a third to a half of the size. A narrow Gaussian also responds fastest on its sides. Right at the top a small shift changes nothing to first order, which is why each difference curve crosses zero near a peak, going negative on the inner side where the neighbours moved away from R_s and positive on the outer side.

With rc of 6 A there is another strong dip near 3.6 A from the 6 atoms of the shell at 3.76 A. The broad eta of 2 curve smears the shells together and its largest change falls to 0.30. So an MLIP sees strain as a pattern of positive and negative lobes at the shell distances, and a small eta with R_s points spaced closer than the Gaussian width picks it up best.

## Exercise 4. $G^4$: separating the angular and radial parts
For a symmetric triplet with bond length $b$ and angle $\theta$, $r_{jk}=2b\sin(\theta/2)$ and
$$G^4=\underbrace{2^{1-\zeta}(1+\lambda\cos\theta)^\zeta}_{\text{angular}}\;\underbrace{e^{-\eta(2b^2+r_{jk}^2)}}_{\text{radial}}\;f_c(b)^2f_c(r_{jk}).$$
Because $r_{jk}$ depends on $\theta$, the radial part also carries angular information. We see how $\eta$ reshapes $G^4(\theta)$ for $b=1.42$ Å (a C–C bond).

```python
import numpy as np, matplotlib.pyplot as plt

fc = lambda r, rc: np.where(r < rc, 0.5 * (np.cos(np.pi * r / rc) + 1), 0.0)
b, rc = 1.42, 6.0
theta = np.linspace(40, 180, 300); t = np.deg2rad(theta)
rjk = 2 * b * np.sin(t / 2)

def G4(zeta, lam, eta):
    return 2**(1 - zeta) * (1 + lam * np.cos(t))**zeta * np.exp(-eta * (2 * b**2 + rjk**2)) * fc(b, rc)**2 * fc(rjk, rc)

fig, ax = plt.subplots(1, 2, figsize=(10, 3.2))
for eta in (0.0, 0.05, 0.2):
    for lam, st in ((1, "-"), (-1, "--")):
        g = G4(1, lam, eta)
        ax[0].plot(theta, g / g.max(), st, label=f"eta={eta}, lambda={lam:+d}")
ax[0].set_xlabel("angle (deg)"); ax[0].set_ylabel("G4 / max"); ax[0].legend(fontsize=7)

g_plus, g_minus = G4(1, 1, 0.05), G4(1, -1, 0.05)
ok = theta < 179                                     # G4(lambda=+1) = 0 at exactly 180 deg
ax[1].plot(theta[ok], g_minus[ok] / g_plus[ok]); ax[1].set_yscale("log")
ax[1].set_xlabel("angle (deg)"); ax[1].set_ylabel("G4(lambda=-1) / G4(lambda=+1)")
plt.tight_layout(); plt.show()

for th in (90.0, 109.5, 120.0, 150.0):
    c = np.cos(np.deg2rad(th))
    ratio = (1 - c) / (1 + c)                           # tan^2(theta/2); diverges only at exactly 180 deg
    print(f"{th:5.1f} deg: ratio = {ratio:6.3f}  -> recovered angle {2 * np.degrees(np.arctan(np.sqrt(ratio))):.1f} deg")
```

**Your turn.** With $\eta=0.2$ the $\lambda=-1$ curve no longer peaks at $180^\circ$ but near $85^\circ$. Explain this shift from the $e^{-\eta r_{jk}^2}$ factor (which angles give a long $r_{jk}$?). Why does the ratio $G^4(\lambda=-1)/G^4(\lambda=+1)$ not depend on $\eta$ at all?

Answer. The radial factor depends on the angle through r_jk = 2 b sin(theta/2). Its contribution to the exponent is minus 4 eta b squared sin squared(theta/2). Write s = sin squared(theta/2) and a = 4 eta b squared. The product of the angular and radial terms then depends on s through s times exp(minus a s), which peaks at s = 1/a.

For eta = 0.2 and b = 1.42 A, a is about 1.61, so s is about 0.62, sin(theta/2) is about 0.79, and theta is about 105 degrees. This is not the 85 degrees mentioned in the notebook note, so check the shift against your plot before relying on either value. The direction of the shift is clear: a larger eta pulls the peak away from 180 degrees toward smaller angles.

The ratio G4(lambda = minus 1) divided by G4(lambda = plus 1) does not depend on eta. The radial factor exp(minus eta r_jk squared), the cutoff functions and the 2 to the power (1 minus zeta) prefactor are identical in numerator and denominator, so they cancel. What remains is the angular ratio (1 - cos theta) divided by (1 + cos theta), which equals tan squared(theta/2). The printed values of 1.000, 2.002, 3.000 and 13.93 match this at 90, 109.5, 120 and 150 degrees, and the recovered angles confirm it.
