```python
test_set=["CCO", "CC(=O)O", "C1CCBr", "c1ccncc1" "OS(=O)(=O)O"]
def heavy_atoms(smiles):
  count={}
  for s in smiles:
    heavy_a_count=0
  i=0
  while i<len(smiles):
    char= s[i]
    if char=='C' and (i+1)<len(s) and s[i+1]=='1':
      heavy_a_count+=1
      i+=2
    elif char=='B' and (i+1)<len(s) and s[i+1]=='r':
        heavy_a_count+=1
        i+=2
    elif char in ['C','c','N','O','S','F','P','I','B']:
        heavy_a_count+=1
        i+=1
    else:
        i+=1
    count[s]=heavy_a_count
    pass
  return count
for idx in test_set:
  print(idx, heavy_atoms(idx))
```
