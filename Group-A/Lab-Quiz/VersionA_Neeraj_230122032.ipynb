```python
def heavy_atoms(smiles):
  counts={}
  i=0;
  while i<len(smiles):
    if smiles[i:i+2] == "Br" or smiles[i:i+2] == "Cl":
      temp = smiles[i:i+2]
      i += 2
    elif smiles[i] in ['C', 'c', 'O', 'o', 'S', 's', 'N', 'n']:
      temp = smiles[i].upper()
      i += 1
    else:
      i += 1
      continue
    if temp in counts:
      counts[temp] += 1
    else:
      counts[temp] = 1

  return counts
```

```python
test_set = ["CCO", "CC(=O)O", "ClCCBr", "c1ccncc1", "OS(=O)(=O)O"]
for ch in test_set:
  print(heavy_atoms(ch))
```

```python
import numpy as np

eps = np.array([[1.0, 0.5],
                [0.2, 1.0]])
A = np.array([1.2, 0.8])

print("Determinant of eps:", np.linalg.det(eps))
c = np.linalg.solve(eps,A)
print("C:",c)
eps_c = np.linalg.matmul(eps,c)
print("eps@c:", eps_c)
if np.array_equal(eps_c,A):
  print("VERIFIED")
else:
  print("NOT VERIFIED")
```

```python
import numpy as np

A_nbutane = np.array([[0,1,0,0], [1,0,1,0], [0,1,0,1], [0,0,1,0]], float)
A_isobutane = np.array([[0,1,1,1], [1,0,0,0], [1,0,0,0], [1,0,0,0]], float)

h = np.ones(4)
sum_agg_nbutane = A_nbutane@h
sum_agg_isobutane = A_isobutane@h
print("sum_agg_nbutane:", sum_agg_nbutane)
print("sum_agg_isobutane:", sum_agg_isobutane)
d1 = np.linalg.norm(sum_agg_isobutane)
d2 = np.linalg.norm(sum_agg_nbutane)
print("d1:", d1)
print("d2:", d2)
mean_agg_nbutane = sum_agg_nbutane/d2
mean_agg_isobutane = sum_agg_isobutane/d1
print("mean_agg_nbutane:", mean_agg_nbutane)
print("mean_agg_isobutane:", mean_agg_isobutane)
```

```python
import numpy as np

T = np.array([300., 320., 340., 360., 380.])
k = np.array([1.71e-3, 6.03e-3, 1.83e-2, 4.92e-2, 1.19e-1])
R = 8.314

ln_k = np.log(k)
inv_T = np.reciprocal(T)

a, b = np.polyfit(ln_k, inv_T, 1)
print("Slope:", a)
print("Intercept:", b)
Ea = (-R*a)/1000
A = np.exp(b)
print("Ea:", Ea, "Kj per mol")
print("A:", A)
print(" The slope carries Activation Energy and the intercept carries A")
```
