```python
def heavy_atoms(smiles_list):
  counts = {}
  for smiles_str in smiles_list:
    i = 0
    while i < len(smiles_str):
      current_char = smiles_str[i]
      next_char = smiles_str[i+1] if i + 1 < len(smiles_str) else ''

      if current_char == 'C' and next_char == 'l':
        counts['Cl'] = counts.get('Cl', 0) + 1
        i += 2
      elif current_char == 'B' and next_char == 'r':
        counts['Br'] = counts.get('Br', 0) + 1
        i += 2
      elif current_char.upper() in "CONS":
        atom_symbol = current_char.upper()
        counts[atom_symbol] = counts.get(atom_symbol, 0) + 1
        i += 1
      else:
        i += 1

  return counts

test_set=['CCO',"CC(=O)O","ClCCBr","c1ccncc1","OS(=O)(=O)O"]
for test in test_set:
  print(test,heavy_atoms(test))
```

```python
import numpy as np

eps = np.array([[1.0,0.5,],[0.2,1.0]])
A = np.array([1.2,0.8])

print("Det",np.linalg.det(eps))
c=np.linalg.solve(eps,A)
print("Solve",c)
d=eps@c
if np.array_equal(d,A):
  print("Verified")
else:
  print("Not verified")
```

because it is numerically stable and cheaper to use solve instead of calculating the inverse

```python
import numpy as np

T=np.array([300.,320.,340.,360.,380.,])
k=np.array([1.71e-3,6.03e-3,1.83e-2,4.92e-2,1.19e-1])
R=8.314

ln_k=np.log(k)
invT=np.reciprocal(T)

a,b = np.polyfit(ln_k,invT,1)
print("Slope",a)
print("intercept",b)
Ea=(-R*a)/1000
A=np.exp(b)
print("Ea", Ea, "Kj per mole")
print("A",A)
print("solpe :  activation energy and intercept : A")
```
