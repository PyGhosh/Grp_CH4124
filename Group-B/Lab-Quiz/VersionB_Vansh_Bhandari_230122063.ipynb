Question 1

```python
reactions = [
    "CC(=O)O.OCC>[H+].[Cl-]>CC(=O)OCC.O",
    "C=C.[H][H]>[Pd]>CC",
    "c1ccccc1.O=[N+]([O-])O>>c1ccccc1[N+](=O)[O-].O",
]
def parse_reaction(rxn):
    f = rxn.split(">")
    if len(f) != 3: raise ValueError("need exactly two '>'")
    return {k: [m for m in x.split(".") if m]
            for k, x in zip(["reactants","reagents","products"], f)}

for r in reactions:
    d = parse_reaction(r)
    print(r); [print(" ", k, len(v), v) for k, v in d.items()]
    print(" ")

print("Reaction 1 is esterification")
print("Empty reagent means no catalyst/solvent was recorded")
```

```python
import numpy as np
E = np.array([
 [2.,0.,-1.,0.],
 [6.,0.,0.,-2.],
 [0.,2.,-2.,-1.]
])

S = np.linalg.svd(E, compute_uv=False)
v = np.linalg.svd(E)[2][-1]
x = v/v[0]; xw = x*2
print(S)
print((S>1e-10).sum())
print(4-(S>1e-10).sum())
print(x)
print(xw)
print(np.round(E@xw))

print()
print("Nullity 1 means the balancing is unique upto to an overall scale factor")
```

```python
import numpy as np
A = np.zeros((4,4))

A = np.diag([1.,1.,1.], 1); A += A.T
w = np.sort(np.linalg.eigvalsh(A))[::-1]
print(w)
print(A.sum(1))
print(2*w[0]+2*w[1])
print(2*w[0]+2*w[1]-4)
print()
print("Butadiene's 0.472β is far smaller than benzene's 2β, so butadiene is only weakly delocalised while benzene is aromatically stabilised.")
```

Question 4

```python
import numpy as np

A6 = np.zeros((6,6))
for i in range(6):
  A6[i, (i+1)%6] = A6[(i+1)%6, i] = 1.0

H = np.random.default_rng(1).normal(size=(6,3))

At = A6 + np.eye(6); P = At/At.sum(1, keepdims=True)
m = np.sort(abs(np.linalg.eigvals(P)))[::-1]
print(m, m[1])
for k in [0,1,2,4,8,16]:
    Hk = np.linalg.matrix_power(P,k)@H
    print(k, np.linalg.norm(Hk-Hk.mean(0), axis=1).max())

print()
print("Deviation decays like μᵏ, so beyond ~3–4 layers atoms become nearly indistinguishable.")
```

```python
import numpy as np

X = np.array([
    [78.1, 2.3],
    [92.1, 1.7],
    [106.2, 2.8],
    [120.2, 2.0],
    [134.2, 3.1],
    [148.2, 2.4]
])
def pca(M):
    C = M.T@M/len(M); w,V = np.linalg.eigh(C)
    i = np.argsort(w)[::-1]; w,V = w[i], V[:,i]
    V *= np.sign(V[abs(V).argmax(0), [0,1]])
    print(C, w, w/w.sum(), V[:,0])


Xc = X - X.mean(0)
pca(Xc)
print()
pca(Xc/Xc.std(0))
print()

print("Molar mass has a variance ~2600× larger than dipole moment, so unstandardised PCA just recovers the mass axis, the standardised result is the one to report.")
```
