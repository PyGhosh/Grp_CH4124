```python
test_set = ["CCO", "CC(=O)O", "ClCCBr", "c1ccncc1", "OS(=O)(=O)O" ]
```

```python
def heavy_atoms(smiles):
  count = {}
  i = 0
  while i < len(smiles):
    char = smiles[i]

    if char.isdigit() or char in '()+-#:':
      i += 1
      continue

    if i + 1 < len(smiles):
      two_char = smiles[i:i+2]
      if two_char == 'Cl':
        element = 'Cl'
        i += 2
      elif two_char == 'Br':
        element = 'Br'
        i += 2
      else:
        element = char.upper()
        i += 1
    else:
      element = char.upper()
      i += 1

    if element not in ['H'] and len(element) == 1 and element.isalpha():
        count[element] = count.get(element, 0) + 1
    elif len(element) == 2 and element.isalpha():
        count[element] = count.get(element, 0) + 1
  return count
```

```python
for smiles_string in test_set:
  atom_cnt = heavy_atoms(smiles_string)
  total_heavy_atoms = sum(atom_cnt.values())
  print(f"SMILES: {smiles_string}")
  print(f"  Atom Counts: {atom_cnt}")
  print(f"  Total Heavy Atoms: {total_heavy_atoms}")
  print(" ")
```

The "Sc" is ambiguos in SMILES because the first character is S which can be interpret as Sulphur and than the c would be considered as Carbon giving thwo different atoms. This is resolved by our notation in which we are first checking if there is an lower character follwed by and uppercharacter then it is considered as single element. (We would have checked the first character of the of two character pairs and if uper is followed by lower we would save the element and increase the i by 2 else increase the i by 1)

```python
import numpy as np

eps = np.array([[1.0, 0.5] , [0.2, 1.0]])
A = np.array([1.2, 0.8])
```

```python
det_eps = np.linalg.det(eps)
print(f"Determinant of eps: {det_eps}")


if det_eps == 0:
  print("The calibration matrix is not invertible. Cannot solve for concentrations.")
else:

  c = np.linalg.solve(eps, A)
  print(f"Concentration vector c: {c}")

  product = eps @ c
  print(f"Product eps @ c: {product}")
```

np.linalg.solve is preferred over np.linalg.inv @ A because it is more numerically stable as it avoids direct computation of the inverse which can introduce inaccuracies.

```python
import numpy as np

A6 = np.zeros((6, 6))
```

```python
for i in range(6):
  A6[i, (i + 1) % 6] = 1
  A6[i, (i - 1) % 6] = 1

ev = np.linalg.eigvals(A6)
sorted_ev = np.sort(ev)
x_k_occupied = sorted_ev[:3]
n_beta = 2 * np.sum(x_k_occupied)
E_pi = n_beta
delocalization_energy = n_beta - 6

print(f"The six eigenvalues of the adjacency matrix are: {sorted_ev}")
print(f"E_pi (as a multiple of beta): {E_pi}")
print(f"Delocalization energy (as a multiple of beta): {delocalization_energy}")
```

The negative delocalization energy (since beta is -ve) indicates that benzene is more stable than three isolated ethylene units, which is a key characteristic of aromatic compounds.

```python
import numpy as np

A_nbutane = np.array([[0,1,0,0],[1,0,1,0], [0,1,0,1], [0,0,1,0]], float)
A_isobutane = np.array([[0,1,1,1],[1,0,0,0], [1,0,0,0], [1,0,0,0]], float)
h = np.ones(4)
```

```python
print("\n n-Butane ")
ev_nb = np.linalg.eigvals(A_nbutane)
sum_ev_nb = np.sum(ev_nb)
mean_ev_nb = np.mean(ev_nb)
sq_sum_ev_nb = np.sum(ev_nb**2)
deg_nb = np.sum(A_nbutane, axis=1)

print(f"EV (nb): {ev_nb}")
print(f"Sum EV (nb): {sum_ev_nb}")
print(f"Mean EV (nb): {mean_ev_nb}")
print(f"Sq-Sum EV (nb): {sq_sum_ev_nb}")
print(f"Deg C (nb): {deg_nb}")

print("\n Isobutane ")
ev_ib = np.linalg.eigvals(A_isobutane)
sum_ev_ib = np.sum(ev_ib)
mean_ev_ib = np.mean(ev_ib)
sq_sum_ev_ib = np.sum(ev_ib**2)
deg_ib = np.sum(A_isobutane, axis=1)

print(f"EV (ib): {ev_ib}")
print(f"Sum EV (ib): {sum_ev_ib}")
print(f"Mean EV (ib): {mean_ev_ib}")
print(f"Sq-Sum EV (ib): {sq_sum_ev_ib}")
print(f"Deg C (ib): {deg_ib}")
```

For n-butane and isobutane, the squared-sum readout successfully separates the isomers due to their different bonding topologies and the degree of carbons represents the number of direct carbon-carbon connections each carbon has.

```python
import numpy as np
T = np.array([300., 320., 340., 360., 380.]) #K
k = np.array([1.71e-3, 6.03e-3, 1.83e-2, 4.92e-2, 1.19e-1])
R = 8.314
```

```python
ln_k = np.log(k)
inv_T = 1 / T


slope, intercept = np.polyfit(inv_T, ln_k, 1)


Ea_J_mol = -slope * R
Ea_kJ_mol = Ea_J_mol / 1000


A = np.exp(intercept)


predicted_ln_k = slope * inv_T + intercept
residuals = ln_k - predicted_ln_k
largest_residual_ln_k = np.max(np.abs(residuals))

print(f"Slope: {slope}")
print(f"Intercept: {intercept}")
print(f"Activation Energy (Ea): {Ea_kJ_mol:.2f} kJ/mol")
print(f"Pre-exponential Factor (A): {A:.2e}")
print(f"Largest Residual in ln(k): {largest_residual_ln_k:.2e}")
```
