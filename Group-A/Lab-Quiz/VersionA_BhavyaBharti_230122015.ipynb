```python
#Q1 a.
test_set = ["CCO", "CC(=O)O", "ClCCBr" , "c1ccncc1", "OS(=O)(=O)O"]
def heavy_atoms(smiles):
  counts ={}
  i=0
  while i<len(smiles):
      if(smiles[i]=='C' and smiles[i+1]=='l'):
        counts['Cl']=counts.get('Cl',0)+1
        i+=2
      elif(smiles[i]=='B' and smiles[i+1]=='r'):
        counts['Br']=counts.get('Br',0)+1
        i+=2
      elif(smiles[i]=='C'):
        counts['C']=counts.get('C',0)+1
        i+=1
      elif(smiles[i]=='O'):
        counts['O']=counts.get('O',0)+1
        i+=1
      elif(smiles[i]=='c'):
        counts['C']=counts.get('C',0)+1
        i+=1
      elif(smiles[i]=='n'):
        counts['N']=counts.get('N',0)+1
        i+=1
      elif(smiles[i]=='N'):
        counts['N']=counts.get('N',0)+1
        i+=1
      elif(smiles[i]=='S'):
        counts['S']=counts.get('S',0)+1
        i+=1
      else:
        i+=1

  return counts

for smile in test_set:
  print(heavy_atoms(smile))

```
