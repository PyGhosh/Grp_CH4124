```python
test_set = ["CCO","CC(=O)O","ClCCBr","c1ccncc1","OS(=O)(=O)O"]

def heavy_atoms(smiles):
  count = {}
  total_heavy_atoms = 0
  i = 0

  while i < len(smiles):
    char = smiles[i]
    if char.isalpha() and char.isupper():
      if i + 1 < len(smiles) and smiles[i+1].islower():
        two_char_atom = smiles[i:i+2]
        if two_char_atom in ['Cl', 'Br']:
          count[two_char_atom] = count.get(two_char_atom, 0) + 1
          total_heavy_atoms += 1
          i += 2
          continue

      count[char] = count.get(char, 0) + 1
      total_heavy_atoms += 1
      i += 1
    elif char.isalpha() and char.islower():
      count[char.upper()] = count.get(char.upper(), 0) + 1
      total_heavy_atoms += 1
      i += 1
    else:
      i += 1
  return count, total_heavy_atoms


for smile in test_set:
    atom_counts, total = heavy_atoms(smile)
    print(f"SMILES: {smile}")
    print(f"  Atom Counts: {atom_counts}")
    print(f"  Total Heavy Atoms: {total}")
    print("-" * 30)


print("Sc is ambiguous in Smiles because S defined as Sulfur and c as carbon then it assume sulfur and carbon rather than Scadium")
```
