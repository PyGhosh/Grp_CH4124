```python
# defining function parse_reaction(rxn) that returns a dictionary with keys reactants, reagents, products
def parse_reaction(rxn):
  # split on ">", check there are three fields, return value error if exactly two > are not present
  fields = rxn.split(">")
  if len(fields) != 3:
    raise ValueError("Reaction must have exactly three fields separated by '>'")
  # split each field on "." and drop empty entries
  reactants = [r.strip() for r in fields[0].split(".") if r.strip()]
  reagents = [r.strip() for r in fields[1].split(".") if r.strip()]
  products = [r.strip() for r in fields[2].split(".") if r.strip()]
  # for every string, print each field with number of molecules it contains and the list itself
  print(f"Reactants: {len(reactants)} molecules")
  print(reactants)
  print(f"Reagents: {len(reagents)} molecules")
  print(reagents)
  print(f"Products: {len(products)} molecules")
  print(products)
  pass
```

```python
for r in reactions:
  parse_reaction(r)
```



1.   Reaction 1, i.e., "CC(=O)O.OCC>[H+].[Cl-]>CC(=O)OCC.O" is an esterification reaction.
2.   The empty reagent field implies that the reactants do not require a catalyst/medium to react.

**Question 04**

```python
A6 = np.zeros((6, 6))
for i in range(6):
  A6[i, (i+1)%6] = A6[(i+1)%6, i] = 1.0
H = np.random.default_rng(1).normal(size=(6,3))
```

```python
A_tidle = A6 + np.eye(6)
print(A_tidle)
```

```python
D_tidle_diag = np.sum(A_tidle, axis=1)
D_tidle = np.diag(D_tidle_diag)
P = np.linalg.inv(D_tidle) @ A_tidle
print(P)
```

```python
# smoothening rate mu
EP = np.sort(abs(np.linalg.eigvals(P)))[::-1]
# print the second largest eigenvalue modulus of P
print(EP[1])
```

```python
# the largest deviation of any atom from the mean row at each value of k = 0, 1, 2, 4, 8, 16
for k in [0,1,2,4,8,16]:
    Hk = np.linalg.matrix_power(P,k)@H
    print(k, np.linalg.norm(Hk-Hk.mean(0), axis=1).max())
```

As the number of layers increases, the deviation reduces, approaching zero, indicating that all the atoms are close to the mean row.

```python
# PCA
def pca(M):
    C = M.T@M/len(M); w,V = np.linalg.eigh(C)
    i = np.argsort(w)[::-1]; w,V = w[i], V[:,i]
    V *= np.sign(V[abs(V).argmax(0), [0,1]])
    print(C, w, w/w.sum(), V[:,0])
```
