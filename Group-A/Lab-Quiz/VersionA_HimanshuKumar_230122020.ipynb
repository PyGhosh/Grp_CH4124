```python
test_set = ["CCO", "CC(=O)O", "ClCCBr", "c1ccncc1", "OS(=O)(=O)O"]
def heavy_atoms(smiles):
  counts = {}
  i = 0
  n = len(smiles)
  while i < n:
    if smiles[i] == 'C' and i != n-1:
      if smiles[i+1] == 'l':
        counts['Cl'] = counts.get('Cl', 0) + 1
        i += 2
      else:
        counts['C'] = counts.get('C', 0) + 1
        i += 1
      continue
    if smiles[i] == 'B':
      counts['Br'] = counts.get('Br', 0) + 1
      i += 2
      continue
    if smiles[i] in ['=', '#', '(' , ')', '1']:
      i += 1
      continue
    if smiles[i] == 'c':
      counts['C'] = counts.get('C', 0) + 1
      i += 1
      continue
    if smiles[i] == 'n':
      counts['N'] = counts.get('N', 0) + 1
      i += 1
      continue
    if smiles[i] == 'o':
      counts['O'] = counts.get('O', 0) + 1
      i += 1
      continue
    counts[smiles[i]] = counts.get(smiles[i], 0) + 1
    i += 1
  return counts

for smile in test_set:
  print(heavy_atoms(smile))


```

```python
eps = np.array([[1.0, 0.5], [0.2, 1.0]])
A = np.array([1.2, 0.8])
det_eps = np.linalg.det(eps)
print("Determinant of eps = " + str(det_eps))
if det_eps == 0 :
  print("Matrix not invertible")
else:
  c = np.linalg.solve(eps, A)
  print("Concentration vector =",c)
  res = eps @ c
  print("Product of eps @ c =", res)
  if np.array_equal(res, A):
    print("Back Substitution done")
```

Solving the equation using the solve function is more beneficial then solving inv function because the solve function takes lesser time complexity as compared to the inv casculation.

```python

def readout(h_agg):
    return np.sum(h_agg)

A_n = np.array([[0,1,0,0], [1,0,1,0], [0,1,0,1], [0,0,1,0]], float)
A_i = np.array([[0,1,1,1], [1,0,0,0], [1,0,0,0], [1,0,0,0]], float)
h = np.ones(4)

d_n = np.sum(A_n, axis=1)
sagg_n = A_n @ h
print("Sum aggregate for n butane = ", sagg_n)
mean_agg_n = sagg_n / d_n
print("Mean Agregate for n butane =", mean_agg_n)

d_i = np.sum(A_i, axis=1)
sagg_i = A_i @ h

print("Sum aggregate for iso butane = ", sagg_i)
mean_agg_i = sagg_i / d_i
print("Mean Agregate for iso butane =", mean_agg_i)

res1 = readout(sagg_n)
res2 = readout(sagg_i)
print("Readout of n butane = ", res1)
print("Readout of iso butane = ", res2)
```
