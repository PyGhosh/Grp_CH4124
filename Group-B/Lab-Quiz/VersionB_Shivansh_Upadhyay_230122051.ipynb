```python
#Q1
reactions = [
    "CC(=O)O.OCC>[H+].[Cl-]>CC(=O)OCC.O",
    "C=C.[H][H]>[Pd]>CC",
    "c1ccccc1.O=[N+]([O-])O>>c1ccccc1[N+](=O)[O-].O",
]
def parse_reaction(rxn):
    f = rxn.split(">")
    if len(f) != 3: raise ValueError("need exactly two '>'")
    return {k: [m for m in x.split(".") if m]
            for k, x in zip(["reactants","reagents","products"], f)}

for r in reactions:
    d = parse_reaction(r)
    print(r); [print(" ", k, len(v), v) for k, v in d.items()]
    print(" ")

print("Reaction 1 is esterification")
print("Empty reagent means no catalyst/solvent was recorded")
```

```python
#Q2
E=np.array([[2,0,-1,0],[6,0,0,-2],[0,2,-2,-1]])

# perform svd
U, s, Vh = np.linalg.svd(E)

# EXTRACT NULL VECTOR
null_vector = Vh[-1, :]
print("Null Vector:", null_vector)

# RESCALE
norm = np.linalg.norm(null_vector)
null_vector_normalized = null_vector / norm
scaled_null_vector = null_vector_normalized * 2
print("Scaled Null Vector:", scaled_null_vector)

# VERIFY E@x=0
verification_result = E @ scaled_null_vector
print("experimental answer:", verification_result)
print("is it close", np.allclose(verification_result, 0))
```

```python
#Q3
import numpy as np

A = np.zeros((4,4))

for i in range(3):
  A[i, (i + 1) ] = 1
  A[(i + 1) , i] = 1

eigenvalues = np.linalg.eigvalsh(A)
eigenvalues.sort()
eigenvalues = eigenvalues[::-1]

E_pi = 2 * eigenvalues[0] + 2 * eigenvalues[1]

delocal_energy_coeff = E_pi - 4

print("Adjacency Matrix (A):\n", A)
print("\nEigenvalues (descending):", eigenvalues)
print(f"\nE_pi: 4*alpha + {E_pi:}*beta")
print(f"Delocalisation Energy: {delocal_energy_coeff:}*beta")
```

```python
#Q4
import numpy as np

A=np.zeros((6, 6))
for i in range(6):
    A[i, (i + 1) % 6] = A[(i + 1) % 6, i] = 1.0

At=A+np.eye(6)
P=At/At.sum(axis=1,keepdims=True)

H=np.random.default_rng(1).normal(size=(6, 3))

v=np.linalg.eigvals(P)
v=np.sort(np.abs(v))[::-1]
mu= v[1]

print("eigenvalue moduli:", v)
print("mu:", mu)

for k in [0, 1, 2, 4, 8, 16]:
    Z = H.copy()
    for _ in range(k):
        Z = P @ Z
    d = np.max(np.linalg.norm(Z - Z.mean(axis=0), axis=1))
    print(k, d)
```

```python
#Q5
import numpy as np

X = np.array([[78.1, 2.3],
              [92.1, 1.7],
              [106.2, 2.8],
              [120.2, 0.7],
              [134.2, 3.1],
              [148.2, 2.4]])

def pca(X):
    Xc = X - X.mean(axis=0)
    C = Xc.T @ Xc / len(X)
    w, v = np.linalg.eigh(C)
    j = np.argsort(w)[::-1]
    w, v = w[j], v[:, j]
    for i in range(v.shape[1]):
        q = np.argmax(np.abs(v[:, i]))
        if v[q, i] < 0:
            v[:, i] *= -1
    return C, w, w / w.sum(), v

C, w, f, v = pca(X)
Z = (X - X.mean(axis=0)) / X.std(axis=0)
Cz, wz, fz, vz = pca(Z)

print("centered covariance:\n", C)
print("eigenvalues:", w)
print("fraction:", f)
print("PC1:", v[:, 0])

print("\nstandardized covariance:\n", Cz)
print("eigenvalues:", wz)
print("fraction:", fz)
print("PC1:", vz[:, 0])
```
