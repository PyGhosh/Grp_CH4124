```python
test_set = ["CCO", "CC(=O)O", "ClCCBr","c1ccncc1","OS(=O)(=O)O"]
def heavy_atoms(smiles):
  counts = {}
  i = 0
  length = len(smiles)
  while i < length:
    if smiles[i] == 'C' and i != length-1:
      if smiles[i+1] == 'l' :
        counts['Cl'] = counts.get('Cl',0) + 1
        i+=2
      else:
        counts['C'] = counts.get['C',0] + 1
        i+=1
        continue
    if smiles[i] == 'B':
        counts[Br]= counts.get('Br',0) + 1
        i+=2
        continue
    if smiles[i] in ['-','=','#',':']:
          i+=1
          continue
    if smiles[i] == 'C':
          counts['C'] = counts.get('C',0) + 1
          i+=1
          continue
    if smiles[i] == 'O':
          counts['O'] = counts.get('O',0) + 1
          i+=1
          continue
    if smiles[i] == 'N':
          counts['N'] = counts.get('N',0) + 1
          i+=1
          continue

    counts[smiles[i]] = counts.get(smiles[i],0) + 1
    i+=1
    continue
    return counts
    for smile in test_set:
      print(heavy_atoms(smiles))

```

```python
eps = np.array([[1.0,0.5],[0.2,1.0]])
A = np.array([1.2,0.8])
det_eps = np.linalg.det(eps)
print("det of matrix " + str(det_eps))
if det_eps == 0:
  print("matrix not invertible")
else:
    c = np. linalg.solve(eps,A)
    print ("matrix",c)
res = eps@c
print("product of eps @ c: " , res)
if np.array_equal(res,A):
  print("verified")

```
